```python
# ### Two Sum Problem
# Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.
# You may assume that each input would have exactly one solution, and you may not use the same element twice.

def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i

# Example usage:
nums = [2,4,5,7,8,9,14]
target = 20
result = two_sum(nums, target)

print(result)  # Output: (0, 1)
```

```python
### Palindrome Number
# Given an integer `x`, return `true` if `x` is a palindrome, and `false` otherwise.
class Solution(object):
    def isPalindrome(self, x):
        # Convert the number to string
        x_str = str(x)
        # Check if the string is equal to its reverse
        return x_str == x_str[::-1]

pal = Solution()

# Example usage:
x = 12345  # True
print(pal.isPalindrome(x))
```

```python
### Palindrome Number using math   
# Given an integer `x`, return `true` if `x` is a palindrome, and `false` otherwise.
class Solution(object):
    def number_to_list_math(self, n):
        if not isinstance(n, int):
            raise TypeError("Input must be an integer.")
        if n < 0:
            n = abs(n) # Handle negative numbers
        if n == 0:
            return [0]
        
        digits = []
        while n > 0:
            last_digit = n % 10
            digits.append(last_digit)
            n = n // 10
    
        digits.reverse()
        return digits

    def isPalindrome(self, n):
        x_list = self.number_to_list_math(n)
        return x_list == x_list[::-1]

pal = Solution()

n = 12345654321
# Example usage:
print("Is", n, "Palindrome?", pal.isPalindrome(n))
```

#### Creating a sub class

```python
# parent class Vehicle
class Vehicle:
  def __init__(self, name, speed):
    self.name = name
    self.speed = speed

# subclass Car that inherits from Vehicle
class Car(Vehicle):
  def __init__(self, name, speed, fuel_type):
    # Using super() to call the constructor of the parent class
    # We can skip the self parameter when using super()
    super().__init__(name, speed)
    self.fuel_type = fuel_type
  def honk_horn(self):
    print("Beep beep!")
# Creating an instance of the Car class
my_car = Car("Toyota", 120, "Gasoline")
print(my_car.name)  # Output: Toyota
print(my_car.speed) # Output: 120
print(my_car.fuel_type) # Output: Gasoline
my_car.honk_horn()  # Output: Beep beep!
```

#### polymorphism 

```python
# Parent class vehicle

class Vehicle:
    def __init__(self, name):
        self.name = name

    def drive(self):
        print(f"The {self.name} is moving.")

# Subclass car that inherits from vehicle
class Car(Vehicle):
    def __init__(self, name, color):
        super().__init__(name)
        self.color = color

    def drive(self):
        print(f"The {self.color} {self.name} is driving on the road.")

vehicle = Vehicle("Bike")
car = Car("Tesla", "red")
vehicle.drive() 
car.drive()
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

class Library(Book):
    def __init__(self):
        self.books = []
        
    def add_book(self, book):
        self.books.append(book)
    
    def display_books(self):
        for b in self.books:
            print(b.title, "-", b.author)

my_library = Library()
book1 = Book("1984", "George Orwell")
book2 = Book("To Kill a Mockingbird", "Harper Lee")

my_library.add_book(book1)
my_library.add_book(book2)

my_library.display_books()
```

```python
## Fibonacci Sequence
def fibonacci(n):
    if n == 0 or n == 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

# Example usage:
n = 10
print(f"Fibonacci of {n} is {fibonacci(n)}")
```

```python
## Fibonacci Sequence
def fibonacci(n):
    memo = {0: 0, 1: 1}  # Base cases
    if n in memo:
        return memo[n]
    else:
        memo[n] = fibonacci(n - 1) + fibonacci(n - 2)
        return memo[n]

# Example usage:
n = 10
print(f"Fibonacci of {n} is {fibonacci(n)}")
```

```python
def fibonacci_inefficient(n):
    print(f"Calculating fibonacci({n})...")
    if n == 0 or n == 1:
        return n
    else:
        return fibonacci_inefficient(n - 1) + fibonacci_inefficient(n - 2)

fibonacci_inefficient(4)
```

```python
memo = {}  # Our memoization dictionary

def fibonacci_memoized(n):
    print(f"Calculating fibonacci({n})...")

    # Step 1: Check if the value is already calculated
    if n in memo:
        print(f"Found fibonacci({n}) in memo, returning stored value.")
        return memo[n]

    # Step 2: Handle base cases and calculate recursively
    if n == 0 or n == 1:
        result = n
    else:
        result = fibonacci_memoized(n - 1) + fibonacci_memoized(n - 2)

    # Step 3: Store the result before returning
    memo[n] = result
    return result

fibonacci_memoized(4)
```

#### The staircase Problem

Imagine you are climbing a staircase with `n` steps. At each step, you can either climb 1 step or 2 steps. You need to find the total number of distinct ways you can climb to the top.

```python
# The staircase problem
# Define a memoization dictionary
memo = {}

# define the recursive function with memoization
def staircase(n):

    # Check if the value is already computed
    if n in memo:
        return memo[n]
    # Base cases and recursive case
    if n == 1 or n == 2:
        return n
    else:
        result = staircase(n - 1) + staircase(n - 2)
        
    # Store the computed value in the memo dictionary
    memo[n] = result
    return result
    
# Example usage:
n = 5   # Number of steps   
print(f"Number of ways to climb {n} steps: {staircase(n)}")
```

```python
# Define a Node class for a singly linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Define a LinkedList class to manage the linked list
class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_the_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def print_list(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")
# Example usage:
ll = LinkedList()  
ll.insert_at_the_beginning(3)
ll.insert_at_the_beginning(2)
ll.insert_at_the_beginning(1)
ll.print_list()  # Output: 1 -> 2 -> 3 -> None
```
